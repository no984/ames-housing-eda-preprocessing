```python
numerical_data['Garage Area']=numerical_data['Garage Area'].fillna(0)
numerical_data["Has_Garage"] = (numerical_data["Garage Area"] > 0).astype(int)
```

```python
x_cat=cat_data[['MS Zoning','Neighborhood','Bsmt Qual','Foundation','Exterior 1st','Exterior 2nd','House Style']]
x_num=numerical_data[['Gr Liv Area','Overall Qual','Garage Area','Year Built','Total Bsmt SF','Has_Garage']]
y=target_values
x=pd.concat([x_cat,x_num],axis=1)
```

#spliting data to train and test

```python
from sklearn.model_selection import train_test_split
X_test,X_train,Y_test,Y_train=train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
)
```

feature engineering

```python

num_feature=x.select_dtypes(include=np.number).columns
cat_feature=x.select_dtypes(exclude=np.number).columns
```

```python
df[(df["Has_Garage"] == 0) & (df["Garage Area"] > 0)]
```

piplines

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

num_pipline=Pipeline([
    ("imputer",SimpleImputer(strategy='median')),
    ('scale',StandardScaler())
])

cat_pipline=Pipeline([
    ('imputer',SimpleImputer(strategy='most_frequent')),
    ('encod',OneHotEncoder(handle_unknown='ignore'))
])
```

#combine every thing

```python
preprocessing=ColumnTransformer([
    ('num',num_pipline,num_feature),
    ('cat',cat_pipline,cat_feature)
])
```

 6. Fit ONLY on training

```python
X_train_prepared = preprocessing.fit_transform(X_train)
X_test_prepared = preprocessing.transform(X_test)
```

```python
print(X_test_prepared)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

model = Pipeline([
    ("preprocessing", preprocessing),
    ("model", LinearRegression())
])

model.fit(X_train, Y_train)

y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(Y_test, y_pred)
rmse = np.sqrt(mean_squared_error(Y_test, y_pred))
r2 = r2_score(Y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
```
